## <a name="pm">Probabilité des mots </a>

Compte tenu du dictionnaire des comptes de mots ou de fréquence, calculons la probabilité que chaque mot apparaisse s'il est choisi au hasard dans le corpus de mots.

$$P(w_i) = \frac{C(w_i)}{M} $$
où

$C(w_i)$ est le nombre total de fois que $w_i$ apparaît dans le corpus.

$M$ est le nombre total de mots dans le corpus.

Par exemple, la probabilité que le mot "am" soit dans la phrase **'I am happy because I am learning'** est :

$$P(am) = \frac{C(w_i)}{M} = \frac {2}{7} .$$

Implémentons la fonction `get_probs` qui nous donne la probabilité
qu'un mot apparaît dans un échantillon. Cela donne un dictionnaire dont les clés sont des mots, et la valeur de chaque mot est sa probabilité dans le corpus de mots.

```python
def get_probs(word_count_dict) :
    '''
    Entrée :
        word_count_dict : Le dictionnaire wordcount où la clé est le mot et la valeur sa fréquence.
    Sortie :
        probs : Un dictionnaire où les clés sont les mots et les valeurs sont la probabilité qu'un mot se produise.
    '''
    probs = {} # renvoie cette variable correctement


    m = sum(word_count_dict.values())
    for key in word_count_dict.keys():
        probs[key] = word_count_dict[key] / m

    return probs
```

```python
probs = get_probs(word_count_dict)
print(f"La longueur de probs est {len(probs)}")
print(f"P('thee') est {probs['thee']:.4f}")
```

## <a name="sup">Suppression </a>

Implément une fonction `delete_letter()` qui, étant donné un mot, renvoie une liste de chaînes de caractères avec un caractère supprimé.

Par exemple, étant donné le mot **nice**, elle retournerait l'ensemble : {'ice', 'nce', 'nic', 'nie'}.

**Étape 1 : Créez une liste de "splits". C'est la seule façon de diviser un mot en deux parties, gauche et droite : Par exemple,   
nice est divisé en : "[(", "nice"), ('n', "ice"), ('ni', "ce"), ('nic', "e"), ('nice', '')]`.
Cette fonction est commune aux quatre fonctions (supprimer, remplacer, permuter, insérer).


```python
def delete_letter(word, verbose=False) :
    '''
    Entrée :
        word : la chaîne/le mot pour lequel vous allez générer tous les mots possibles
                dans le vocabulaire qui ont 1 caractère manquant
    Sortie :
        delete_l : une liste de toutes les chaînes possibles obtenues en supprimant 1 caractère du mot
    '''

    delete_l = []
    split_l = []


    for c in range(len(word)):
        split_l.append((word[:c],word[c :]))
    for a,b in split_l:
        delete_l.append(a+b[1 :])


    if verbose: print(f"mot en entrée: {word}, \nsplit_l = {split_l}, \ndelete_l = {delete_l}")

    return delete_l
```

## <a name="cha">Changement</a>

Maintenant, implémentons une fonction `switch_letter()` qui permet de changer deux lettres dans un mot. Elle prend un mot et renvoie une liste de tous les commutateurs possibles de deux lettres **qui sont adjacents l'un à l'autre**.
- Par exemple, pour le mot "eta", elle renvoie {"eat", "tea"}, mais ne renvoie pas "ate".

**Etape 1:** est la même que dans delete_letter()  
**Etape 2:** Compréhension d'une liste ou d'une boucle qui forme des chaînes de caractères en échangeant les lettres adjacentes. Ceci est de la forme :  
f(L,R) pour L, R en fractionnement si condition]` où "condition" testera la longueur de R dans une itération donnée. Voir ci-dessous

```python
def switch_letter(word, verbose=False):
    '''
    Entrée :
        word : chaîne de saisie
     Sortie :
        switches : une liste de toutes les chaînes possibles avec un caractère adjacent commuté
    '''

    switch_l = []
    split_l = []

    len_word=len(word)
    for c in range(len_word):
        split_l.append((word[:c],word[c:]))
    switch_l = [a + b[1] + b[0] + b[2:] for a,b in split_l if len(b) >= 2]


    if verbose: print(f"Mot d'entrée = {word} \nsplit_l = {split_l} \nswitch_l = {switch_l}")

    return switch_l
```

## <a name="rem">Remplacement </a>

 Implémentons maintenant une fonction `replace_letter()` qui prend un mot et renvoie une liste de chaînes de caractères avec une **lettre de remplacement** du mot original.

**Etape 1:** est la même que dans `delete_letter()`

**Etape 2:** Une liste de compréhension ou pour les boucles qui forment des chaînes en remplaçant des lettres.  Cela peut être de la forme :  
f(a,b,c) pour a, b en fractionnement si condition pour c en chaîne]` Notez l'utilisation de la seconde pour boucle.  
Dans cette routine, on s'attend à ce qu'un ou plusieurs des remplacements incluent le mot original. Par exemple, en remplaçant la première lettre de "ear" par "e", on obtient "ear".

**Etape 3:** Supprimons la lettre d'entrée originale de la sortie.

```python
def replace_letter(word, verbose=False) :
    '''
    Entrée :
        word : la chaîne de saisie/le mot
    Sortie :
        replaces : une liste de toutes les chaînes de caractères possibles où nous avons remplacé une lettre du mot original.
    '''

    letters = "abcdefghijklmnopqrstuvwxyz"
    replace_l = []
    split_l = []


    for c in range(len(word)):
        split_l.append((word[0:c],word[c:]))
    replace_l = [a + l + (b[1:] if len(b)> 1 else '') for a,b in split_l if b for l in letters]
    replace_set=set(replace_l)
    replace_set.remove(word)

    # transformer l'ensemble en une liste et la trier, pour une visualisation plus facile
    replace_l = sorted(list(replace_set))

    if verbose: print(f"Mot d'entrée = {word} \nsplit_l = {split_l} \nreplace_l {replace_l}")

    return replace_l
```

## <a name="ins">Insertion</a>

**Instructions pour insert_letter()** : Implémentons maintenant une fonction `insert_letter()` qui prend un mot et retourne une liste avec une lettre insérée à chaque décalage.

**Etape 1:** est la même que dans `delete_letter()`

**Etape 2:** Il peut s'agir d'une liste de compréhension du formulaire :  
`[f(a,b,c) pour a, b en fractionnement (splits) si condition pour c en chaîne de caractère]`.   

```python
def insert_letter(word, verbose=False):
    '''
    Entrée :
        word : la chaîne de saisie/le mot
    Sortie :
        inserts : un ensemble de toutes les chaînes possibles avec une nouvelle lettre insérée à chaque décalage
    '''
    letters = 'abcdefghijklmnopqrstuvwxyz'
    insert_l = []
    split_l = []


    for c in range(len(word)+1):
        split_l.append((word[0:c],word[c:]))
    insert_l = [ a + l + b for a,b in split_l for l in letters]

    if verbose: print(f"Mot d'entrée = {word} \nsplit_l = {split_l} \ninsert_l = {insert_l}")

    return insert_l
```

# <a name="3">III. Combiner les modifications</a>

Maintenant que nous avons mis en œuvre les manipulations de chaînes, nous allons créer deux fonctions qui, à partir d'une chaîne, renverront toutes les modifications ou éditions simples et doubles possibles sur cette chaîne. Il s'agira de `edit_one_letter()` et de `edit_two_letters()`.

## <a name="mp">  Une modification près</a>

Implémentons la fonction `edit_one_letter` pour obtenir toutes les modifications possibles qui sont à une modification près d'un mot. Les modifications consistent à remplacer, insérer, effacer et, éventuellement, à changer d'opération. Nous devons utiliser les fonctions précédentes que nous avons déjà implémentées pour compléter cette fonction. La fonction "switch" est une fonction d'édition moins courante, donc son utilisation sera sélectionnée par un argument d'entrée "allow_switches".

Notez que ces fonctions renvoient des *listes* alors que cette fonction devrait renvoyer un *ensemble python* (set). L'utilisation d'un ensemble élimine toute entrée en double.

```python
def edit_one_letter(word, allow_switches = True):
    """
    Entrée :
        word : la chaîne/le mot pour lequel nous allons générer tous les mots possibles qui sont à une édition près.
    Sortie :
        edit_one_set : un ensemble de mots avec une seule modification possible. Veuillez renvoyer un ensemble et non une liste.
    """

    edit_one_set = set()


    edit_one_set.update(delete_letter(word))
    if allow_switches:
        edit_one_set.update(switch_letter(word))
    edit_one_set.update(replace_letter(word))
    edit_one_set.update(insert_letter(word))


    return edit_one_set
```

```python
tmp_word = "at"
tmp_edit_one_set = edit_one_letter(tmp_word)
# transformer ceci en liste pour la trier, afin de la visualiser
tmp_edit_one_l = sorted(list(tmp_edit_one_set))

print(f"Mot en entrée {tmp_word} \nedit_one_l \n{tmp_edit_one_l}\n")
print(f"Le type de l'objet retourné doit être un ensemble {type(tmp_edit_one_set)}")
print(f"Le nombre de sorties de edit_one_letter('at') est {len(edit_one_letter('at'))}")
```

Nous pouvons maintenant généraliser cette méthode à mettre en œuvre pour obtenir deux modifications sur un mot. Pour ce faire, nous devons obtenir toutes les modifications possibles sur un seul mot et ensuite, pour chaque mot modifié, nous devons le modifier à nouveau.



## <a name="dmp">  Deux modification près</a>


Implémentons la fonction `edit_two_letters` qui retourne un ensemble de mots qui sont à deux éditions de distance. Notons que la création d'éditions supplémentaires basées sur la fonction `edit_one_letter` peut 'restaurer' certaines éditions à zéro ou une seule. Cela est autorisé ici. Ceci est pris en compte dans get_corrections.

```python
def edit_two_letters(word, allow_switches = True):
    '''
    Input:
        word: la chaîne de chaine/mot
    Output:
        edit_two_set: un ensemble de chaînes de caractères avec toutes les deux éditions possibles
    '''

    edit_two_set = set()


    edit_one = edit_one_letter(word,allow_switches=allow_switches)
    for w in edit_one:
        if w:
            edit_two = edit_one_letter(w,allow_switches=allow_switches)
            edit_two_set.update(edit_two)


    return edit_two_set
```

```python
tmp_edit_two_set = edit_two_letters("a")
tmp_edit_two_l = sorted(list(tmp_edit_two_set))
print(f"Nombre de chaînes de caractères avec une distance d'édition de deux : {len(tmp_edit_two_l)}")
print(f"10 premières chaînes {tmp_edit_two_l[:10]}")
print(f"Les 10 dernières chaînes {tmp_edit_two_l[-10 :]}")
print(f"Le type de données de l'objet retourné doit être un ensemble {type(tmp_edit_two_set)}")
print(f"Le nombre de chaînes de caractères qui sont à deux distances d'édition de 'at' est {len(edit_two_letters('at'))}")
```

## <a name="ps">Probabilité de suggestion</a>
Implémentons `get_corrections`, qui renvoie une liste de zéro à n tuples de suggestion possibles de la forme (mot, probabilité_de_mot).

**Etape 1:** Générons des suggestions pour un mot fourni : Nous utiliserons les fonctions d'édition que nous avons développées. L'"algorithme de suggestion" doit suivre cette logique :
* Si le mot est dans le vocabulaire, suggérons le mot.
* Sinon, s'il y a des suggestions de `edit_one_letter` qui sont dans le vocabulaire, utilisons-les.
* Sinon, s'il y a des suggestions de `edit_two_letters` qui sont dans le vocabulaire, utilisons-les.
* Sinon, suggérons le mot d'entrée.*  
* L'idée est que les mots générés à partir de moins d'éditions sont plus probables que les mots avec plus d'éditions.


Note :
- Les modifications d'une ou deux lettres peuvent "restaurer" des chaînes de caractères à zéro ou à une seule modification. Cet algorithme tient compte de ce fait en sélectionnant de préférence les éditions à faible distance en premier.

En Python, les opérations logiques telles que "et" et "ou" ont deux propriétés utiles. Elles peuvent fonctionner sur des listes et elles ont un comportement de ["court-circuit"](https://docs.python.org/3/library/stdtypes.html). Essayons-les :

```python
# exemple d'opération logique sur des listes ou des ensembles
print( [] and ["a","b"] )
print( [] or ["a","b"] )
#exemple de comportement en cas de court-circuit
val1 =  ["Most","Likely"] or ["Less","so"] or ["least","of","all"]  # sélectionne en premier, n'évalue pas le reste
print(val1)
val2 =  [] or [] or ["least","of","all"] # continue l'évaluation jusqu'à ce qu'il y ait une liste non vide
print(val2)
```

Le "ou" logique pourrait être utilisé pour mettre en œuvre l'algorithme de suggestion de manière très compacte. Alternativement, les constructions "si/alors" pourraient être utilisées.

**Etape 2** : Créons un dictionnaire des "meilleurs mots" dans lequel la "clé" est une suggestion et la "valeur" est la probabilité que ce mot figure dans votre vocabulaire. Si le mot n'est pas dans le vocabulaire, attribuez-lui une probabilité de 0.

**Étape 3** : Sélectionnons les n meilleures suggestions. Il peut y en avoir moins que n.

```python
def get_corrections(word, probs, vocab, n=2, verbose = False):
    '''
    Entrée :
        word : une chaîne de caractères saisie par l'utilisateur pour vérifier les suggestions
        probs : un dictionnaire qui associe chaque mot à sa probabilité dans le corpus
        vocabulaire : un ensemble contenant tout le vocabulaire
        n : nombre de corrections de mots possibles que vous souhaitez voir figurer dans le dictionnaire
    Sortie :
        n_best : une liste de tuples avec les n mots corrigés les plus probables et leurs probabilités.
    '''

    suggestions = []
    n_best = []


    suggestions = list((word in vocab and word) or edit_one_letter(word).intersection(vocab) or edit_two_letters(word).intersection(vocab))
    n_best = [[s,probs[s]] for s in list(reversed(suggestions))]


    if verbose: print("suggestions = ", suggestions)

    return n_best
```

Nous mettrons en place un système de programmation dynamique qui nous indiquera le nombre minimum de modifications nécessaires pour convertir une chaîne de caractères en une autre.

## <a name="pd"> Programmation dynamique</a>

La programmation dynamique décompose un problème en sous-problèmes qui peuvent être combinés pour former la solution finale. Ici, étant donné une chaîne source [0..i] et une chaîne cible [0..j], nous allons calculer toutes les combinaisons de sous-chaînes [i, j] et calculer leur distance d'édition. Pour faire cela efficacement, nous utiliserons un tableau pour maintenir les sous-chaînes calculées précédemment et utiliserons celles-ci pour calculer des sous-chaînes plus grandes.

Vous devez créer une matrice et mettre à jour chaque élément de la matrice comme suit :  

$$\text{Initialization}$$

\begin{align}
D[0,0] &= 0 \\
D[i,0] &= D[i-1,0] + del\_cost(source[i]) \\
D[0,j] &= D[0,j-1] + ins\_cost(target[j]) \\
\end{align}



$$\text{Per Cell Operations}$$
\begin{align}
 \\
D[i,j] =min
\begin{cases}
D[i-1,j] + del\_cost\\
D[i,j-1] + ins\_cost\\
D[i-1,j-1] + \left\{\begin{matrix}
rep\_cost; & if src[i]\neq tar[j]\\
0 ; & if src[i]=tar[j]
\end{matrix}\right.
\end{cases}
\end{align}

\begin{align}
 \\
D[i,j] =min
\begin{cases}
D[i-1,j] + del\_cost\\
D[i,j-1] + ins\_cost\\
D[i-1,j-1] + \left\{\begin{matrix}
rep\_cost; & if src[i]\neq tar[j]\\
0 ; & if src[i]=tar[j]
\end{matrix}\right.
\end{cases}
\end{align}

The variable `sub_cost` (for substitution cost) is the same as `rep_cost`; replacement cost.  We will stick with the term "replace" whenever possible.

```python
def min_edit_distance(source, target, ins_cost = 1, del_cost = 1, rep_cost = 2):
    '''
    Entrée :
        source : une chaîne de caractères correspondant à la chaîne de caractères avec laquelle vous commencez.
        target : une chaîne de caractères correspondant à la chaîne de caractères avec laquelle vous voulez finir
        ins_cost : un nombre entier définissant le coût d'insertion
        del_cost : un nombre entier définissant le coût de suppression
        rep_cost : un nombre entier définissant le coût de remplacement.
    Sortie :
        D : une matrice de len(source)+1 par len(cible)+1 contenant les distances d'édition minimales.
        med : la distance d'édition minimale (med) requise pour convertir la chaîne source en chaîne cible
    '''

    # utiliser les coûts de suppression et d'insertion comme 1
    m = len(source)
    n = len(target)
    #initialiser la matrice des coûts avec des zéros et des dimensions (m+1,n+1)
    D = np.zeros((m+1, n+1), dtype=int)



    # Remplir la colonne 0, de la ligne 1 à la ligne m, toutes deux incluses
    for row in range(1,m+1): # Remplacer None par la plage appropriée
        D[row,0] = D[row-1,0] + del_cost

    # Remplir la rangée 0, pour toutes les colonnes de 1 à n, toutes deux incluses
    for col in range(1,n+1): # Remplacer None par la plage appropriée
        D[0,col] = D[0,col-1] + ins_cost

    # Boucle sur la ligne 1 à la ligne m, toutes deux incluses
    for row in range(1,m+1):

        # Boucle de la colonne 1 à la colonne n, toutes deux incluses
        for col in range(1,n+1):

           # Initialiser r_cost au coût 'replace' qui est passé dans cette fonction
            r_cost = rep_cost

            # Vérifier si le caractère source de la ligne précédente
            # correspond au caractère cible de la colonne précédente,
            if source[row-1] == target[col-1]:
                # Mettre à jour le coût de remplacement à 0 si la source et la cible sont identiques
                r_cost = 0

            # Mettre à jour le coût à la ligne, à la colonne en fonction des entrées précédentes dans la matrice des coûts.
            # Se référer à l'équation calculer pour D[i,j] (le minimum des trois coûts calculés)
            D[row,col] = min([D[row-1,col]+del_cost, D[row,col-1]+ins_cost, D[row-1,col-1]+r_cost])

     # Définir la distance d'édition minimale avec le coût trouvé à la ligne m,
    med = D[m,n]

    return D, med
```

```python
source =  'play'
target = 'stay'
matrix, min_edits = min_edit_distance(source, target)
print("minimum edits: ",min_edits, "\n")
idx = list('#' + source)
cols = list('#' + target)
df = pd.DataFrame(matrix, index=idx, columns= cols)
print(df)
```
